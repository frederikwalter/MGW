# Human–Mouse Alignment with merged bins (HESTA → MOSTA)

This notebook is `human_mouse_align.ipynb` with one change: a section is reduced to `MAX_OBS` points by merging neighbouring bins instead of keeping a random subset (see *Subsampling by merging bins* below).

We align a sagittal section of a human embryo from HESTA (https://db.cngb.org/hesta/download/) to a sagittal section of a mouse embryo from MOSTA (Chen et al. '22) with MGW. The human section is the source A and the mouse section the target B, so the coupling `P` is HESTA × MOSTA and carries mouse labels and expression onto the human section.

**Which sections.** Neither atlas records where a section lies along the left–right axis. The laterality run (`laterality_test/laterality_score.ipynb`, results in `laterality/`) scores each section from 0 (midline) to 1 (far lateral) and pairs every HESTA section with the MOSTA section of the matched stage whose score is closest (`best`). **Pick the HESTA section in the next cell**; its MOSTA partner is read from `laterality/laterality_pairs.csv` unless `MOSTA_SECTION` is set.

**Control.** Cross-species differences only mean something relative to the differences between two individuals of the same species. The notebook therefore also aligns the HESTA section with a section of *another* HESTA embryo of the same stage (closest laterality score) and reports both runs side by side.

**Features.** `FEATURE_MODE` selects what MGW sees:
- `"native"`: each species keeps its own genes and PCA ("multimodal", as in `mouse_align.ipynb`). A CCA maps the two feature spaces onto each other. It is fit on pairs of bins matched by the feeler, which by default uses the orthologs (`FEELER_PAIRING`, see *Why orthologs* in section 2).
- `"orthologs"`: both sections are embedded in one space spanned by one-to-one orthologs (Ensembl BioMart, table in `mgw/resources/`, `orthologs.shared_embedding`), so no CCA is needed.

The orthologs also evaluate the coupling (co-expression check below). The evaluation genes are left out of the shared embedding.

## Subsampling by merging bins

The GW solve needs each section reduced to about `MAX_OBS` points. `human_mouse_align.ipynb` keeps a random subset of the bins and discards the rest. Here `SUBSAMPLE` chooses how a section is reduced (`util.merge_bins`):
- `"random"`: the bins `human_mouse_align.ipynb` keeps (same seed). Nothing is merged.
- `"voronoi"`: the same bins serve as seeds, and every other bin joins its nearest seed. `LLOYD_ITER` Lloyd steps then move each seed to the centre of its tile, which evens out the tile sizes. With `LLOYD_ITER = 0` the only difference from `"random"` is the merging.
- `"grid"`: square tiles, with the side chosen so that about `MAX_OBS` tiles contain tissue.

A merged point sits at the mean position of its bins and carries their majority organ class. `purity` is the share of that class, `n_bins` the number of bins.

`AGG` sets how the counts of a merged point are combined:
- `"sum"`: the summed counts, which is what Stereo-seq does for its coarser bin sizes. The mean of the counts gives the same features, because `normalize_total` removes the per-point scale.
- `"rms"`: the root mean square of the counts. It gives more weight to single high-count bins, on the idea that this compensates for dropouts. With the same total, a gene seen in one bin of a tile scores higher than a gene spread over all its bins.

All points get the same mass in the GW problem, as in `human_mouse_align.ipynb`.

A section with at most `MAX_OBS` bins is not reduced, so these settings only matter for larger sections. With `MAX_OBS = 14_000`, the early stages are reduced little or not at all (CS12-15 and E9.5-E10.5 have about 5-37k bins). From CS17 (about 66k bins) and CS18 (130-150k) on, each point merges about 5-10 bins. To study merging on small sections, lower `MAX_OBS`.

**Comparing strategies.** Scores on merged points cannot be compared with scores on random bins, because the points differ. Every bin of the full section is therefore assigned to a point: its tile, or for `"random"` its nearest kept bin. It takes that point's transferred class and transported expression. Label accuracy and ortholog co-expression are then scored on all bins (`*_full` in the summary). Each run appends its rows to `human_mouse_subsample_summary.csv`, so running the notebook once per setting builds up the comparison.

## Pair selection from the laterality run

`laterality_pairs.csv` lists, for every HESTA section, the MOSTA sections of the matched stage(s) (notes, Table 1) ranked by `abs_dL`, the difference in laterality score. The HESTA score is `L_pair`, the score smoothed by the valley fit along the section order of the embryo; the fit pools neighbouring sections to a common value, so a difference of exactly 0 can come from pooling. The raw scores `L_raw` are printed as well.

`orientation = "test both"` means the side of the MOSTA section is unknown, so the pair may be mirror images. MGW compares only distances within each section and cannot tell a section from its mirror image; running both orientations would return the same coupling. The **mirror test** further down checks instead which orientation the labels and the ortholog expression support.

```python
# Stage correspondence quality (notes, Table 1)
STAGE_QUALITY = {
    ("CS12-13", "E9.5"): "good", ("CS14-15", "E10.5"): "good",
    ("CS17", "E11.5"): "ambiguous", ("CS17", "E12.5"): "ambiguous",
    ("CS18", "E12.5"): "good",
    ("CS19", "E12.5"): "ambiguous", ("CS19", "E13.5"): "ambiguous",
    ("CS20", "E13.5"): "good", ("CS23", "E14.5"): "weak",
}

pairs = pd.read_csv(LATERALITY_DIR / "laterality_pairs.csv")
scores = pd.read_csv(LATERALITY_DIR / "laterality_scores.csv", dtype={"embryo": str}).set_index("key")
fits = pd.read_csv(LATERALITY_DIR / "laterality_hesta_fits.csv", dtype={"embryo": str})

def key(dataset, sec):
    return f"HESTA:{sec}_HESTA" if dataset == "HESTA" else f"MOSTA:{sec}.MOSTA"

def name(k):
    return k.split(":", 1)[1].removesuffix("_HESTA").removesuffix(".MOSTA")

h_key = key("HESTA", HESTA_SECTION)
cand = pairs[(pairs.HESTA == h_key) & pairs.MOSTA.notna()].sort_values("abs_dL")
if cand.empty:
    raise ValueError(f"{HESTA_SECTION} has no MOSTA candidate in {LATERALITY_DIR / 'laterality_pairs.csv'}")
m_key = cand[cand.best].MOSTA.iloc[0] if MOSTA_SECTION is None else key("MOSTA", MOSTA_SECTION)
MOSTA_SEC = name(m_key)
if m_key not in scores.index:
    raise KeyError(f"{MOSTA_SEC} has no laterality score")

h, m = scores.loc[h_key], scores.loc[m_key]
pair_row = cand[cand.MOSTA == m_key]
fit = fits[(fits.stage == h.stage) & (fits.embryo == h.embryo)]
pair_info = {
    "HESTA": HESTA_SECTION, "MOSTA": MOSTA_SEC,
    "stages": f"{h.stage} / {m.stage}",
    "stage match": STAGE_QUALITY.get((h.stage, m.stage), "not in Table 1"),
    "L_pair (HESTA / MOSTA)": f"{h.L_pair:.3f} / {m.L_pair:.3f}",
    "L_raw (HESTA / MOSTA)": f"{h.L_laterality:.3f} / {m.L_laterality:.3f}",
    "abs_dL": abs(h.L_pair - m.L_pair),
    "HESTA side": h.side,
    "orientation": pair_row.orientation.iloc[0] if len(pair_row) else "n/a (manual pair)",
    "embryo fit": (f"{fit['shape'].iloc[0]}, rms {fit.fit_rms.iloc[0]:.3f}, p {fit.p_value.iloc[0]:.3f}"
                   if len(fit) else "n/a"),
}
print(pd.Series(pair_info).to_string())
print("\nMOSTA candidates for", HESTA_SECTION)
print(cand[["MOSTA", "L_HESTA", "L_MOSTA", "abs_dL", "orientation", "best"]].head(6).round(3).to_string(index=False))

# Control: a section of another HESTA embryo of the same stage, closest L_pair
CONTROL = CONTROL_SECTION
if RUN_CONTROL and CONTROL is None:
    hs = scores[(scores.dataset == "HESTA") & scores.L_pair.notna()]
    others = hs[(hs.stage == h.stage) & (hs.embryo != h.embryo)]
    if len(others):
        CONTROL = name((others.L_pair - h.L_pair).abs().idxmin())
if RUN_CONTROL and CONTROL is None:
    print(f"\nNo control: {h.stage} has no second HESTA embryo.")
elif RUN_CONTROL:
    c = scores.loc[key("HESTA", CONTROL)]
    print(f"\nControl: {CONTROL} (stage {c.stage}, embryo {c.embryo}, side {c.side}), "
          f"L_pair {c.L_pair:.3f}, |dL| to {HESTA_SECTION} = {abs(c.L_pair - h.L_pair):.3f}")

check_files(("HESTA", HESTA_SECTION), ("MOSTA", MOSTA_SEC), *([("HESTA", CONTROL)] if RUN_CONTROL and CONTROL else []))
```

## Helpers

**Loading.** Each section is read, the counts are moved to `X` and `DROP_LABELS` are removed. The full section (counts, coordinates, organ classes) is kept for the evaluation, and `util.merge_bins` reduces it to `MAX_OBS` points. `full.obs["tile"]` holds the point each bin belongs to. Normalization works per point, so it runs after merging. The control reuses the HESTA section.

**Features.** `make_features` log-normalizes copies of two sections. It picks the evaluation genes (the `N_EVAL_GENES` most variable 1:1 orthologs, without mitochondrial genes) and returns their log-normalized expression on both sections. The shared embedding goes to `obsm["X_ortho"]`: the `N_HVG` most variable matched genes of each section without the evaluation genes, z-scored within each section, then a joint PCA. The features MGW uses go to `obsm["X_pca"]`: the PCA of each section in `"native"` mode, the shared embedding in `"orthologs"` mode and for two HESTA sections. `full_expression` computes the same log-normalized expression on every bin of a full section.

```python
def load_section(dataset, sec):
    """Full section (DROP_LABELS removed) and its reduction to MAX_OBS points; full.obs['tile'] maps bins to points."""
    adata = sc.read_h5ad(section_path(dataset, sec))
    counts, label_col = ("counts", "celltype") if dataset == "HESTA" else ("count", "annotation")
    # HESTA X is log-normalized, MOSTA X is something else; both keep raw counts in a layer
    adata.X = adata.layers.pop(counts)
    keep = ~adata.obs[label_col].astype(str).str.lower().isin([l.lower() for l in DROP_LABELS]).to_numpy()
    adata = adata[keep]

    # Keep only what the pipeline needs; coordinates may be stored as uint32
    full = ad.AnnData(sp.csr_matrix(adata.X, dtype=np.float32),
                      obs=pd.DataFrame(index=[f"{sec}_{n}" for n in adata.obs_names]),
                      var=pd.DataFrame(index=adata.var_names.astype(str)))
    full.obsm["spatial"] = np.asarray(adata.obsm["spatial"][:, :2], dtype=np.float64)
    full.obs["label"] = adata.obs[label_col].astype(str).to_numpy()
    full.obs["label_h"], unknown = harmonize(full.obs["label"])
    del adata
    gc.collect()

    method = SUBSAMPLE if MAX_OBS else "random"
    out, tile = util.merge_bins(full, MAX_OBS or full.n_obs, method=method, agg=AGG, label_key="label_h",
                                lloyd_iter=LLOYD_ITER)
    full.obs["tile"] = tile
    if method != "random":
        out.obs_names = f"{sec}_" + out.obs_names
    for a in (out, full):
        a.uns["dataset"], a.uns["section"] = dataset, sec
    nb = out.obs["n_bins"]
    how = method if method == "random" else f"{method}, {AGG}"
    print(f"[{sec}] {full.n_obs:,} bins (of {keep.size:,}, {keep.size - keep.sum():,} dropped) x {full.n_vars:,} genes"
          f" -> {out.n_obs:,} points ({how}); bins per point {nb.mean():.1f} (min {nb.min()}, max {nb.max()}),"
          f" mean purity {out.obs['purity'].mean():.3f}")
    if unknown:
        print(f"  labels not in HARMONIZE (scored as Other): {unknown}")
    return out, full


def full_expression(full, genes):
    """Log-normalized expression of `genes` on every bin of a full section (normalize_total + log1p)."""
    idx = full.var_names.get_indexer(genes)
    assert (idx >= 0).all(), "evaluation genes missing in the full section"
    tot = np.asarray(full.X.sum(1)).ravel()
    target = np.median(tot[tot > 0])  # as sc.pp.normalize_total
    return dense((sp.diags(target / np.where(tot > 0, tot, 1)) @ full.X[:, idx]).log1p())


def gene_var(X):
    mean = np.asarray(X.mean(0)).ravel()
    sq = np.asarray(X.multiply(X).mean(0)).ravel() if sp.issparse(X) else (X ** 2).mean(0)
    return sq - mean ** 2


def dense(X):
    return np.asarray(X.toarray() if sp.issparse(X) else X, dtype=np.float32)


def make_features(A_raw, B_raw, mode, eval_genes=None):
    """Log-normalize copies of A and B, set obsm['X_pca'] and obsm['X_ortho'], and return the evaluation expression."""
    A, B = A_raw.copy(), B_raw.copy()
    for a in (A, B):
        sc.pp.normalize_total(a)
        sc.pp.log1p(a)
    cross = B.uns["dataset"] == "MOSTA"
    if cross:
        gh, gb = orthologs.restrict_to_orthologs(A.var_names, B.var_names, ORTHO)
    else:
        gh = gb = np.intersect1d(A.var_names, B.var_names)
    XA, XB = A[:, gh].X, B[:, gb].X   # matched columns, named by the human symbols gh

    if eval_genes is None:
        ok = ~pd.Index(gh).str.upper().str.startswith("MT-")
        rank = (rankdata(-gene_var(XA)) + rankdata(-gene_var(XB))) / 2
        rank[~ok] = np.inf
        eval_genes = np.asarray(gh)[np.argsort(rank)[:N_EVAL_GENES]]

    # Shared embedding over the matched genes, without the evaluation genes
    EA, EB, emb_genes = orthologs.shared_embedding(A, B, ORTHO if cross else None, n_hvg=N_HVG,
                                                   n_comps=PCA_COMP, exclude=eval_genes)
    A.obsm["X_ortho"], B.obsm["X_ortho"] = EA, EB
    shared = f"{len(emb_genes):,} {'1:1 orthologs' if cross else 'shared genes'}"
    if cross and mode == "native":
        for a in (A, B):
            sc.pp.pca(a, n_comps=PCA_COMP, random_state=0)
        how = f"per-section PCA on all genes ({A.n_vars:,} human, {B.n_vars:,} mouse); X_ortho on {shared}"
    else:
        A.obsm["X_pca"], B.obsm["X_pca"] = EA, EB
        how = f"shared embedding (X_ortho) on {shared}"
    print(f"[features] {A.uns['section']} vs {B.uns['section']}: {how}")
    idx = pd.Index(gh).get_indexer(eval_genes)
    if (idx < 0).any():
        print(f"  {(idx < 0).sum()} evaluation genes missing in this pair; skipped")
    ev = dict(genes=np.asarray(eval_genes)[idx >= 0], Y_A=dense(XA[:, idx[idx >= 0]]), Y_B=dense(XB[:, idx[idx >= 0]]),
              n_shared=len(gh))
    return A, B, ev
```

**MGW and diagnostics.** `preprocess` and `align` wrap `mgw_preprocess` and `mgw_align_core` with the settings of `hesta_align.ipynb`. `preprocess` passes the distances between the `X_ortho` embeddings as the cost between the sections (`cross_key`), and uses the CCA feeler only for `"native"` features of two species. `align` solves fused GW with the cost between the sections (`pre["M"]`, weight `FUSED_ALPHA`). It reports the GW energy of the final coupling (both cost matrices are scaled to a maximum of 1, so values are comparable between runs) and the linear term ⟨M, P⟩ separately and then frees the n × n matrices, so the control run fits in memory. The plotting and orientation functions are those of `hesta_align.ipynb`.

**Evaluation.**
- *Label transfer*: B's organ classes are moved to A through `P`. Scored are A bins whose class also occurs in B (a class missing in B cannot be predicted); `Other` or `Excluded` predictions count as wrong. Balanced accuracy is the mean recall over these classes.
- *Ortholog co-expression*: for each evaluation gene, the Pearson correlation over A's bins between its human expression and the mouse (or control) expression transported through `P`, $\hat Y = \mathrm{diag}(P\mathbf 1)^{-1} P Y_B$. The baseline transports through `P` after shuffling B's bins, which keeps the marginals but destroys the spatial correspondence. The evaluation genes are not in the shared embedding, but they correlate with genes that are, so label accuracy is the more independent measure.
- *Full resolution*: every bin of A's full section takes the transferred class and the transported expression of its point (`full.obs["tile"]`). Both measures are then computed over all bins, so runs with different `SUBSAMPLE` / `AGG` settings are comparable.
- *Mirror test*: A is rotated, with and without a reflection, onto B (both centred and scaled to unit RMS radius), and every A bin takes its nearest B bin. The best rotation and the best reflection are scored by the same two measures.

```python
def preprocess(A, B):
    """The CCA feeler maps per-species features ("native", two species); shared features need no map."""
    native = FEATURE_MODE == "native" and A.uns["dataset"] != B.uns["dataset"]
    return mgw.mgw_preprocess(
        A, B, PCA_comp=PCA_COMP, CCA_comp=CCA_COMP, use_cca_feeler=native,
        use_pca_X=True, use_pca_Z=True, log1p_X=True, log1p_Z=True,
        verbose=True, feature_only=False, spatial_only=True, rep_norm="zscore",
        cross_key="X_ortho", feeler_pairing=FEELER_PAIRING, feeler_alpha=FEELER_ALPHA, coords=COORDS)


def free_memory():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()


def gw_energy(C1, C2, P):
    """GW objective of P in float32, on the GPU if it fits."""
    def run(dev):
        C1t, C2t, Pt = (torch.as_tensor(np.asarray(x), dtype=torch.float32, device=dev) for x in (C1, C2, P))
        p, q = Pt.sum(1), Pt.sum(0)
        return float(p @ (C1t * C1t) @ p + q @ (C2t * C2t) @ q - 2 * ((C1t @ Pt) * (Pt @ C2t)).sum())
    if device == "cuda":
        try:
            return run("cuda")
        except torch.cuda.OutOfMemoryError:
            free_memory()
    return run("cpu")


def align(pre, tag):
    out = mgw.mgw_align_core(
        pre, widths=PHI_ARC, lr=LR, niter=N_ITER, knn_k=KNN_K, geodesic_eps=GEODESIC_EPS,
        save_dir=None, tag=tag, verbose=True, plot_net=PLOT_NET, gw_params=GW_PARAMS,
        seed=SEED, deterministic=True, n_restarts=1, fused_alpha=FUSED_ALPHA)
    res = dict(P=np.asarray(out["P"]), xs=out["xs"], xs2=out["xs2"])
    res["gw_energy"] = gw_energy(out["C_M"], out["C_N"], res["P"])
    res["fgw_linear"] = float((pre["M"].astype(np.float64) * res["P"]).sum())
    print(f"GW energy {res['gw_energy']:.4e}, linear term <M,P> {res['fgw_linear']:.4f} (FUSED_ALPHA {FUSED_ALPHA})")
    del out
    free_memory()
    return res


def plot_coupling_lines(xa, xb, P, title, names, n_lines=1500, seed=0):
    """Draw B in the frame the coupling implies (reflection allowed); join random A bins to their peak B target."""
    P = np.asarray(P)
    xb_aligned, _, _, _, _ = align_vis.coupling_procrustes(xa, xb, P, ensure_rotation=False)
    idx = np.random.default_rng(seed).choice(len(xa), min(len(xa), n_lines), replace=False)
    jdx = P[idx].argmax(1)
    shift = np.array([xa[:, 0].max() - xb_aligned[:, 0].min() + 0.3 * np.ptp(xa[:, 0]), 0.0])

    fig, ax = plt.subplots(figsize=(11, 5))
    ax.scatter(*xa.T, s=1, c="C0", label=names[0])
    ax.scatter(*(xb_aligned + shift).T, s=1, c="C1", label=f"{names[1]} (coupling frame)")
    ax.add_collection(LineCollection(np.stack([xa[idx], xb_aligned[jdx] + shift], axis=1),
                                     colors="k", alpha=0.08, lw=0.5))
    ax.set_title(f"{title}: peak target of {len(idx):,} random {names[0]} bins")
    ax.set_aspect("equal"); ax.axis("off")
    ax.legend(markerscale=8, frameon=False)
    plt.show()


def orientation_report(xa, xb, P):
    """Rigid fit of A onto its barycentric image in B, as a rotation and as a reflection. Returns det."""
    P = np.asarray(P)
    Y = (P @ xb.astype(P.dtype)) / P.sum(1, keepdims=True)
    Xc, Yc = xa - xa.mean(0), Y - Y.mean(0)
    U, _, Vt = np.linalg.svd(Xc.T @ Yc)
    d = np.sign(np.linalg.det(U @ Vt))
    print(f"orientation det = {d:+.0f}")
    for kind, Q in [("rotation", U @ np.diag([1, d]) @ Vt), ("reflection", U @ np.diag([1, -d]) @ Vt)]:
        rms = np.sqrt(((Xc @ Q - Yc) ** 2).sum(1).mean())
        angle = np.degrees(np.arctan2(Q[0, 1], Q[0, 0]))
        what = f"angle {angle:+4.0f}°" if kind == "rotation" else f"mirror axis {angle / 2:+4.0f}°"
        print(f"best {kind:<10}  {what}  rms {rms:.4f}")
    return int(d)


def row_entropy(p):
    p = p / p.sum()
    return -(p * np.log(p + 1e-30)).sum()


def feeler_report(pre, names):
    feeler = pre["feeler"]
    if feeler is None:
        print("no CCA feeler: the features are already shared")
        return dict(feeler_det=np.nan, feeler_entropy=np.nan, feeler_acc=np.nan)
    ia, ib = feeler["meta"]["idxX_feeler"], feeler["meta"]["idxZ_feeler"]
    xa, xb = feeler["xs"][ia], feeler["xt"][ib]
    P_f = np.asarray(feeler["P_feeler"])
    plot_coupling_lines(xa, xb, P_f, f"CCA feeler coupling ({feeler['meta']['pairing']})", names)
    det = orientation_report(xa, xb, P_f)
    H = np.mean([row_entropy(p) for p in P_f])
    print(f"mean row entropy {H:.2f}  (uniform {np.log(P_f.shape[1]):.2f}, one-to-one 0)")
    lt_f = label_transfer(P_f, pre["A_"][ia], pre["B_"][ib])
    print(f"feeler label transfer: accuracy {lt_f['acc']:.3f}, balanced accuracy {lt_f['bal_acc']:.3f}")
    return dict(feeler_det=det, feeler_entropy=H, feeler_acc=lt_f["acc"])


# ---------- evaluation
def scored_classes(a_true, b_labels):
    return sorted((set(a_true) & set(b_labels)) - NOT_SCORED)


def label_transfer(P, A, B):
    a_true, b_lab = A.obs["label_h"].to_numpy(str), B.obs["label_h"].to_numpy(str)
    pred, _, _ = plotting.project_labels_via_P(P.T, b_lab, direction="A_to_B")
    classes = scored_classes(a_true, b_lab)
    mask = np.isin(a_true, classes)
    recall = {c: float((pred[a_true == c] == c).mean()) for c in classes}
    return dict(pred=np.asarray(pred, dtype=str), classes=classes, mask=mask,
                acc=float((pred[mask] == a_true[mask]).mean()), bal_acc=float(np.mean(list(recall.values()))),
                recall=recall)


def colwise_r(X, Y):
    Xc, Yc = X - X.mean(0), Y - Y.mean(0)
    den = np.sqrt((Xc ** 2).sum(0) * (Yc ** 2).sum(0))
    return np.where(den > 0, (Xc * Yc).sum(0) / np.where(den > 0, den, 1), np.nan)


def coexpression(P, ev, seed=0):
    rowsum = P.sum(1, keepdims=True)
    r = colwise_r(ev["Y_A"], (P @ ev["Y_B"]) / rowsum)
    perm = np.random.default_rng(seed).permutation(len(ev["Y_B"]))
    r0 = colwise_r(ev["Y_A"], (P @ ev["Y_B"][perm]) / rowsum)
    return r, r0


def rigid_scan(A, B, ev, step=5):
    """Nearest-neighbour maps for rotations of A onto B, with and without a reflection."""
    def unit(xy):
        xy = xy - xy.mean(0)
        return xy / np.sqrt((xy ** 2).sum(1).mean())
    a, b = unit(A.obsm["spatial"]), unit(B.obsm["spatial"])
    tree = cKDTree(b)
    a_true, b_lab = A.obs["label_h"].to_numpy(str), B.obs["label_h"].to_numpy(str)
    mask = np.isin(a_true, scored_classes(a_true, b_lab))
    rows = []
    for flip in (False, True):
        xy0 = a * [1, -1] if flip else a
        for deg in range(0, 360, step):
            t = np.radians(deg)
            nn = tree.query(xy0 @ np.array([[np.cos(t), np.sin(t)], [-np.sin(t), np.cos(t)]]))[1]
            rows.append(dict(map="reflection" if flip else "rotation", angle=deg,
                             acc=(b_lab[nn][mask] == a_true[mask]).mean(),
                             median_r=np.nanmedian(colwise_r(ev["Y_A"], ev["Y_B"][nn]))))
    return pd.DataFrame(rows)


def full_scores(P, A_full, B, lt, ev, Y_full, seed=0):
    """Label transfer and ortholog co-expression on every bin of A's full section."""
    tile = A_full.obs["tile"].to_numpy()
    true, pred = A_full.obs["label_h"].to_numpy(str), lt["pred"][tile]
    classes = scored_classes(true, B.obs["label_h"].to_numpy(str))
    mask = np.isin(true, classes)
    rowsum = P.sum(1, keepdims=True)
    perm = np.random.default_rng(seed).permutation(len(ev["Y_B"]))
    r = colwise_r(Y_full, ((P @ ev["Y_B"]) / rowsum)[tile])
    r0 = colwise_r(Y_full, ((P @ ev["Y_B"][perm]) / rowsum)[tile])
    return dict(acc=float((pred[mask] == true[mask]).mean()),
                bal_acc=float(np.mean([(pred[true == c] == c).mean() for c in classes])),
                median_r=float(np.nanmedian(r)), median_r0=float(np.nanmedian(r0)), n_bins=len(tile))


def print_full_scores(fs, names):
    print(f"full resolution, {fs['n_bins']:,} {names[0]} bins: label accuracy {fs['acc']:.3f}, "
          f"balanced {fs['bal_acc']:.3f}; median ortholog r {fs['median_r']:.3f} (shuffled {fs['median_r0']:.3f})")


def mirror_report(scan, lt, r):
    best = []
    for kind, g in scan.groupby("map", sort=False):
        a, e = g.loc[g.acc.idxmax()], g.loc[g.median_r.idxmax()]
        best.append(dict(map=kind, angle_by_labels=a.angle, acc=a.acc, angle_by_expr=e.angle, median_r=e.median_r))
    best = pd.DataFrame(best).set_index("map")
    print(best.round(3).to_string())
    print(f"MGW coupling: acc {lt['acc']:.3f}, median r {np.nanmedian(r):.3f}")
    by_lab = best.acc.idxmax()
    by_expr = best.median_r.idxmax()
    print(f"labels favour the {by_lab}, ortholog expression favours the {by_expr}")
    return best


def plot_label_panels(A, B, pred, names):
    a_lab, b_lab = A.obs["label_h"].to_numpy(str), B.obs["label_h"].to_numpy(str)
    _, _, classes = align_vis.coarsen_to_shared_annotations(a_lab, b_lab, max_classes=8)
    palette = dict(zip([c for c in classes if c != "Other"], align_vis.CATEGORY_COLORS))
    palette["Other"] = "#CBD5E1"
    coarse = lambda lab: np.where(np.isin(lab, list(palette)), lab, "Other")
    fig, ax = plt.subplots(1, 3, figsize=(15, 4.5), constrained_layout=True)
    panels = [(B.obsm["spatial"], b_lab, f"{names[1]}: annotated"),
              (A.obsm["spatial"], a_lab, f"{names[0]}: annotated"),
              (A.obsm["spatial"], pred, f"{names[0]}: transferred from {names[1]}")]
    for a, (xy, lab, title) in zip(ax, panels):
        # dot area ∝ 1/n, so sparse sections look as filled as the 14k-bin ones
        a.scatter(*xy.T, c=[palette[l] for l in coarse(lab)], s=float(np.clip(28_000 / len(xy), 2, 30)), linewidths=0)
        a.set_title(title); a.set_aspect("equal"); a.axis("off")
    fig.legend(handles=[Line2D([0], [0], marker="o", ls="", color=palette[c], label=c) for c in palette],
               loc="outside lower center", ncol=len(palette), frameon=False)
    plt.show()


def plot_confusion(A, lt, title):
    a_true = A.obs["label_h"].to_numpy(str)[lt["mask"]]
    pred = lt["pred"][lt["mask"]]
    cols = lt["classes"] + sorted(set(pred) - set(lt["classes"]))
    M = pd.crosstab(pd.Categorical(a_true, lt["classes"]), pd.Categorical(pred, cols),
                    normalize="index", dropna=False)
    fig, ax = plt.subplots(figsize=(0.55 * len(cols) + 2.5, 0.5 * len(lt["classes"]) + 1.5))
    im = ax.imshow(M.to_numpy(), cmap="Blues", vmin=0, vmax=1)
    ax.set_xticks(range(len(cols)), cols, rotation=60, ha="right")
    ax.set_yticks(range(len(lt["classes"])), lt["classes"])
    ax.set_xlabel("transferred class"); ax.set_ylabel("annotated class")
    ax.set_title(title)
    fig.colorbar(im, ax=ax, label="fraction of annotated class")
    plt.show()


def plot_coexpression(r, r0, title):
    fig, ax = plt.subplots(figsize=(6, 3.2))
    bins = np.linspace(-0.3, 1, 53)
    ax.hist(r0[np.isfinite(r0)], bins=bins, color="#CBD5E1", label=f"shuffled B (median {np.nanmedian(r0):.3f})")
    ax.hist(r[np.isfinite(r)], bins=bins, color="#146C94", alpha=0.8, label=f"coupling (median {np.nanmedian(r):.3f})")
    ax.set_xlabel("Pearson r over A bins"); ax.set_ylabel("genes"); ax.set_title(title)
    ax.legend(frameon=False)
    plt.show()


def plot_merge(red, full):
    """Bins per point, purity, and nearest-neighbour spacing against a random subset of the same size."""
    xy = red.obsm["spatial"]
    fig, ax = plt.subplots(1, 3, figsize=(15, 4), constrained_layout=True)
    for a, key, cmap, title in [(ax[0], "n_bins", "viridis", "bins per point"),
                                (ax[1], "purity", "magma", "purity of the majority class")]:
        im = a.scatter(*xy.T, c=red.obs[key], s=1, cmap=cmap, linewidths=0)
        fig.colorbar(im, ax=a, shrink=0.8)
        a.set_title(f"{red.uns['section']}: {title}"); a.set_aspect("equal"); a.axis("off")
    rnd = full.obsm["spatial"][np.random.default_rng(0).choice(full.n_obs, red.n_obs, replace=False)]
    nn = lambda p: cKDTree(p).query(p, k=2)[0][:, 1]
    d, d0 = nn(xy), nn(rnd)
    bins = np.linspace(0, np.quantile(np.r_[d, d0], 0.995), 60)
    ax[2].hist(d0, bins=bins, color="#CBD5E1", label=f"random bins (CV {d0.std() / d0.mean():.2f})")
    ax[2].hist(d, bins=bins, color="#146C94", alpha=0.8, label=f"{red.uns['merge']['method']} (CV {d.std() / d.mean():.2f})")
    ax[2].set_xlabel("distance to the nearest point"); ax[2].set_ylabel("points"); ax[2].legend(frameon=False)
    plt.show()


def summary_row(run, names, kind, lt, r, r0, mirror, feeler, fs):
    return {
        "run": kind, "A": names[0], "B": names[1], "feature_mode": FEATURE_MODE if kind == "cross-species" else "joint",
        "n_A": len(run["xs"]), "n_B": len(run["xs2"]), "gw_energy": run["gw_energy"], "fgw_linear": run["fgw_linear"],
        "label_acc": lt["acc"], "label_bal_acc": lt["bal_acc"], "n_classes": len(lt["classes"]),
        "coexpr_median_r": float(np.nanmedian(r)), "coexpr_shuffled_r": float(np.nanmedian(r0)),
        "n_A_bins": fs["n_bins"], "label_acc_full": fs["acc"], "label_bal_acc_full": fs["bal_acc"],
        "coexpr_median_r_full": fs["median_r"], "coexpr_shuffled_r_full": fs["median_r0"],
        "det": run["det"], **feeler,
        "rot_acc": mirror.loc["rotation", "acc"], "refl_acc": mirror.loc["reflection", "acc"],
        "rot_r": mirror.loc["rotation", "median_r"], "refl_r": mirror.loc["reflection", "median_r"],
    }
```

# 2. Check the CCA feeler alignment.

**Why orthologs.** The CCA is fit on pairs of bins from a small OT problem between the two sections. The original feeler (`FEELER_PAIRING = "spatial"`) is a GW on spatial distances. GW compares distances within each section only, and a section and its mirror image have the same distances. A spatial-only feeler therefore cannot tell top from bottom; which orientation it returns depends on the subsampling and the solver. A cost between the sections breaks the tie. `"cross"` pairs bins by the distance of their ortholog embeddings (`X_ortho`); `"fused"` adds that cost, weighted by `FEELER_ALPHA`, to the spatial GW. `COORDS = "isotropic"` keeps the aspect ratio of each section, which `"unit_square"` distorts. The final MGW solve uses the same cost, weighted by `FUSED_ALPHA` (section 3).

The plot draws B in the frame the feeler coupling implies (reflection allowed). The orientation check fits a rotation and a reflection to it, the row entropy measures how spread out it is, and the label accuracy scores the feeler pairs the same way as the final coupling. In `"orthologs"` mode there is no feeler.

```python
pre = preprocess(A, B)
A, B = pre["A_"], pre["B_"]
feeler_stats = feeler_report(pre, names)
```

# 3. MGW alignment

φ and ψ are fit to the features of each section, and their pullback metrics give geodesic distances within each section. The coupling solves fused GW: the GW term on these distances plus `FUSED_ALPHA` times the ortholog cost ⟨M, P⟩. GW alone keeps the coupling faithful to the shape of each section; the cost between the sections decides which way round the sections match.

# Mirror test: compare the coupling with the best rigid maps.

This uses the annotations and the orthologs, so it is a diagnostic, not part of MGW. If the reflection beats the rotation (or vice versa) by both measures, the pair has that orientation; MGW agrees if its `det` matches (+1 rotation, −1 reflection) and its scores are close to the better rigid map. Rigid maps ignore the differences in shape between species, so their absolute scores are lower bounds.

```python
scan = rigid_scan(A, B, ev)
mirror = mirror_report(scan, lt, r)

fig, ax = plt.subplots(1, 2, figsize=(11, 3), sharex=True)
for kind, g in scan.groupby("map", sort=False):
    ax[0].plot(g.angle, g.acc, label=kind)
    ax[1].plot(g.angle, g.median_r, label=kind)
ax[0].axhline(lt["acc"], color="k", ls="--", lw=1, label="MGW")
ax[1].axhline(np.nanmedian(r), color="k", ls="--", lw=1, label="MGW")
ax[0].set_ylabel("label accuracy"); ax[1].set_ylabel("median ortholog r")
for a in ax:
    a.set_xlabel("rotation angle (°)")
ax[0].legend(frameon=False)
plt.show()
```

# 4. Control: the same human section against another human embryo

Same pipeline and the same `SUBSAMPLE` / `AGG`, with both sections in the human gene space (joint PCA over the shared genes). The co-expression check uses the same evaluation genes as above, so both runs are scored on the same genes.

```python
rows = [summary_row(run, names, "cross-species", lt, r, r0, mirror, feeler_stats, fs)]
couplings = {"cross-species": (names, run, A, B, A_full, B_full)}

if RUN_CONTROL and CONTROL:
    C_raw, C_full = load_section("HESTA", CONTROL)
    A_c, C_c, ev_c = make_features(A_raw, C_raw, "joint", eval_genes=ev["genes"])
    names_c = (HESTA_SECTION, CONTROL)
    pre_c = preprocess(A_c, C_c)
    A_c, C_c = pre_c["A_"], pre_c["B_"]
    feeler_c = feeler_report(pre_c, names_c)
    run_c = align(pre_c, tag=f"HH_{HESTA_SECTION}_{CONTROL}_{SUB_TAG}_seed{SEED}")
    del pre_c
    free_memory()

    plot_coupling_lines(run_c["xs"], run_c["xs2"], run_c["P"], "MGW coupling (control)", names_c)
    run_c["det"] = orientation_report(run_c["xs"], run_c["xs2"], run_c["P"])
    lt_c = label_transfer(run_c["P"], A_c, C_c)
    print(f"label transfer {CONTROL} → {HESTA_SECTION}: accuracy {lt_c['acc']:.3f}, "
          f"balanced accuracy {lt_c['bal_acc']:.3f} over {len(lt_c['classes'])} classes")
    plot_label_panels(A_c, C_c, lt_c["pred"], names_c)
    r_c, r0_c = coexpression(run_c["P"], ev_c)
    plot_coexpression(r_c, r0_c, f"{len(ev_c['genes'])} genes, {CONTROL} → {HESTA_SECTION}")
    fs_c = full_scores(run_c["P"], A_full, C_c, lt_c, ev_c, full_expression(A_full, ev_c["genes"]))
    print_full_scores(fs_c, names_c)
    mirror_c = mirror_report(rigid_scan(A_c, C_c, ev_c), lt_c, r_c)
    rows.append(summary_row(run_c, names_c, "control (HESTA)", lt_c, r_c, r0_c, mirror_c, feeler_c, fs_c))
    couplings["control (HESTA)"] = (names_c, run_c, A_c, C_c, A_full, C_full)
else:
    print("control skipped")
```

# Summary

One row per alignment. The rows are appended to `human_mouse_subsample_summary.csv` in the alignments folder, with the pair and the main settings, so runs with other sections, feature modes and subsampling settings add up to one table. Compare settings by the `*_full` columns.

```python
summary = pd.DataFrame(rows)
meta = {"date": datetime.now().isoformat(timespec="seconds"),
        "stages": pair_info["stages"], "stage_match": pair_info["stage match"],
        "abs_dL": pair_info["abs_dL"], "HESTA_side": pair_info["HESTA side"],
        "orientation": pair_info["orientation"], "max_obs": MAX_OBS,
        "subsample": SUBSAMPLE if MAX_OBS else "none", "lloyd_iter": LLOYD_ITER if SUBSAMPLE == "voronoi" else 0,
        "agg": AGG if SUBSAMPLE != "random" else "none", "n_iter": N_ITER,
        "drop_labels": ";".join(DROP_LABELS), "seed": SEED,
        "feeler_pairing": FEELER_PAIRING if FEATURE_MODE == "native" else "none",
        "feeler_alpha": FEELER_ALPHA if FEATURE_MODE == "native" and FEELER_PAIRING == "fused" else np.nan,
        "coords": COORDS, "fused_alpha": FUSED_ALPHA}
summary = summary.assign(**meta)
print(summary.drop(columns=list(meta)).set_index("run").T.to_string())
if SUMMARY_CSV.exists():
    # columns may have been added since the file was written; older runs used the spatial feeler,
    # unit-square coords and pure GW
    old = pd.read_csv(SUMMARY_CSV)
    old = old.assign(**{c: v for c, v in {"feeler_pairing": "spatial", "coords": "unit_square", "fused_alpha": 0.0}.items()
                        if c not in old})
    summary = pd.concat([old, summary], ignore_index=True)
summary.to_csv(SUMMARY_CSV, index=False)
print("appended to", SUMMARY_CSV)
```

# Save the couplings for later analysis.

`P` rows and columns refer to the points, so their names are saved with it. `tile_A` / `tile_B` give, for every bin of the full sections (`bins_A` / `bins_B`), the point it belongs to, so `P` can be lifted back to the bins.

```python
for kind, ((na, nb), res, a, b, a_full, b_full) in couplings.items():
    suffix = FEATURE_MODE if kind == "cross-species" else "control"
    path = SAVE_DIR / f"P_{na}_{nb}_{suffix}_{SUB_TAG}.npz"
    np.savez(path, P=res["P"].astype(np.float32), obs_A=a.obs_names.to_numpy(str), obs_B=b.obs_names.to_numpy(str),
             bins_A=a_full.obs_names.to_numpy(str), tile_A=a_full.obs["tile"].to_numpy(),
             bins_B=b_full.obs_names.to_numpy(str), tile_B=b_full.obs["tile"].to_numpy())
    print("Saved", path)
```
